```python
import pandas as pd

file_path = "/mnt/d/all_data/20260909/AY-2526 EMIS ADW GIS (6-8) 226633.csv"

df = pd.read_csv(file_path)

print(df.head())
print(df.columns)
```

```python
df = df[["aadhaar_no", "student_name", "application_id","scholarship_amount","mobile_no"]]
```

```python
df["benificiary_id"]=df["aadhaar_no"]
df["installment_id"]="1"
df["benificiary_reference_id"]=df["application_id"]
df["benificiary_name"]=df["student_name"]
df["benificiary_mobile_no"]=df["mobile_no"]
df["adhaar_no"]=df["aadhaar_no"]
df["amount"]=df["scholarship_amount"]
df["payment_mode_id"]="1"
df["district_code"]="1000"
df["taluk_code"]="1000"
df["jurisdiction_layer_code"]=None
df["jurisdiction_code"]=None
df["benificiary_account_no"]=None
df["benificiary_bank_name"]=None
df["benificiary_ifsc_code"]=None
df["benificiary_iin_no"]=None
```

```python
columns = [
    "benificiary_id",
    "installment_id",
    "benificiary_reference_id",
    "benificiary_name",
    "benificiary_mobile_no",
    "adhaar_no",
    "amount",
    "payment_mode_id",
    "district_code",
    "taluk_code",
    "jurisdiction_layer_code",
    "jurisdiction_code",
    "benificiary_account_no",
    "benificiary_bank_name",
    "benificiary_ifsc_code",
    "benificiary_iin_no"
]

df = df[columns]
```

```python
for col in df.columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.replace(r"\s+", "", regex=True)
        .str.strip()
    )
```

```python
for col in df.columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.replace(r"[^a-zA-Z0-9]", "", regex=True)
    )
```

```python
df = df.fillna("").astype(str)

for col in df.columns:
    df[col] = df[col].str.replace(r"\s+", "", regex=True)
```

```python
df.to_csv("/mnt/d/all_data/20260909/adw_gis_ben_data_6_9.csv", index=False)
```
