```python
import pandas as pd 
```

```python
file_path = "/mnt/d/all_data/20260909/create GBM ID.xlsx"

df = pd.read_excel(file_path, engine="openpyxl")
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", "_", regex=True)
    .str.replace(r"[^a-zA-Z0-9_]", "", regex=True)
)
```

```python
df["aadhaar_number"]=df["aadhar_number"]
df["beneficiary_first_name"]=df["beneficiary_name"]
df["gender"]=df["gender"]
df["date_of_birth"]= df["dob"]
df["community"] = df["community"]
df["mobile_number"] = df["mob_no"]
df["address_line_1"] = df["address_1"]
df["district_id"] = df["district_id"]


colums= [
    "aadhaar_number",
    "beneficiary_first_name",
    "gender",
    "date_of_birth",
    "community",
    "mobile_number",
    "address_line_1",
    "district_id"
]

df = df[colums]
```

```python
df["date_of_birth"] = pd.to_datetime(
    df["date_of_birth"],
    format="%d.%m.%Y",
    errors="coerce"
).dt.strftime("%Y-%m-%d")
```

```python
df.to_csv("/mnt/d/all_data/20260909/twosgbm.csv", index=False)
```
