```python
import pandas as pd
```

```python
file_name = "/data/Data/AY-2526 EMIS 11th pfms_id 165354.csv"
```

```python
df = pd.read_csv(file_name)
```

```python
column = [
    "aadhaar_no",
    "application_id",
    "pfms_id",
    "student_name",
    "pfms_beneficiary_type",
    "district_code",
    "gender"
]
df = df[column]
```

```python
df["gender"] = df["gender"].str.lower().map({
    "female": "F",
    "male": "M"
})
```

```python
for col in df.columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.replace(r"\s+", "", regex=True)
        .str.strip()
    )
```

```python
for col in df.columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.replace(r"[^a-zA-Z0-9]", "", regex=True)
    )
```

```python
df = df.fillna("").astype(str)

for col in df.columns:
    df[col] = df[col].str.replace(r"\s+", "", regex=True)
```

```python
df.to_csv("/data/Data/todeletepfms_id.csv", index=False)
```
