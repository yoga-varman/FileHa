```python
import pandas as pd
filepath = "/home/yogavarman/Projects/FileHandling/Data/extracted_data.csv"
```

```python
df = pd.read_csv(filepath)
```

```python
df.columns = df.columns.str.replace(" ", "", regex=False)
df.columns = df.columns.str.lower()
```

```python
cols = [
    "fullnameinenglish",
    "aadhaarnumber",
    "schemespecificid",
    "statesharepaymentamount"
    ]
df = df[cols]
```

```python
rename_cols = {
    "aadhaarnumber": "aadhaar_number",
    "schemespecificid": "application_id",
    "statesharepaymentamount": "state_share_amount",
    "fullnameinenglish":"beneficiary_name"
    
}
df = df.rename(columns=rename_cols)
```

```python
pfms_df = pd.DataFrame({
    "aadhaar_number": [
        "290184432810",
        "335424444902",
        "666044709287",
        "786002248358",
        "846442347851"
    ],
    "pfms_id": [
        "DTNPE00174631",
        "DTNKC00172382",
        "DTNNG00395990",
        "DTNCHP00103445",
        "DTNPU00282096"
    ]
})

```

```python
df["aadhaar_number"] = df["aadhaar_number"].astype(str)
pfms_df["aadhaar_number"] = pfms_df["aadhaar_number"].astype(str)

df = df.merge(
    pfms_df[["aadhaar_number", "pfms_id"]],
    on="aadhaar_number",
    how="left"
)
```

```python
df["scheme_id"] = "5"
df["scheme_name_identifier"] = "6"
df["academic_year"] = "25-26"
df["registration_type"] = "R"

df["unique_credit_transaction_id"] = (
    df["aadhaar_number"].astype(str)
    + df["pfms_id"].astype(str) + "20260916"
)

df["center_share_amount"] = "0.00"
df["total_amount"] = df["state_share_amount"]
df["district_lgt_code"] = "568"
df["state_lgt_code"] = "33"
df["reference_id"] = "PAY20260916in"
```
